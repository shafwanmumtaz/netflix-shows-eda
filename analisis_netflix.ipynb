```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('data/netflix_titles_sample.csv')
df.head()
```

## 1. Pembersihan Nilai Hilang (Missing Values) & Ekstraksi Fitur Durasi

```python
# Mengisi missing values pada kolom kategorikal
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

# Memisahkan angka durasi
df['duration_value'] = df['duration'].str.extract(r'(\d+)').astype(int)
print('Proporsi Tipe Tayangan:')
print(df['type'].value_counts())
```

## 2. Analisis Distribusi Negara & Rating Usia

```python
print('Distribusi Negara Produsen:')
print(df['country'].value_counts())

print('\nDistribusi Rating Tayangan:')
print(df['rating'].value_counts())
```
